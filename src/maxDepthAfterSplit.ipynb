```python
# CODE

class Solution:
    def maxDepthAfterSplit(self, seq: str) -> list[int]:
        arr = []
        count = 0
        
        for a in seq:
            if a == '(':
                arr.append(count % 2)
                count += 1
            else:
                count -= 1
                arr.append(count % 2)
        
        return arr
    # END
```

```python
seq = "(()())"

Solution().maxDepthAfterSplit(seq)
```
