```python
# CODE

class Solution:
    def isValid(self, s: str) -> bool:
        stack: list[str] = []
        
        bracket: dict[str:str] = {
            '(' : ')',
            '[' : ']',
            '{' : '}'
        }
        
        for a in s:
            if a in bracket:
                stack.append(a)
            else:
                if stack:
                    if a == bracket[stack[-1]]:
                        stack.pop()
                    else:
                        return False
                else:
                    return False
        if stack:
            return False
        return True
    # END
```

```python
s = "()[]{}"

Solution().isValid(s)
```
